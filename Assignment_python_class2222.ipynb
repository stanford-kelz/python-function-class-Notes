# Q1. (8 marks)
# A cinema charges tickets based on age:
#   - Under 5         → Free
#   - 5 to 12         → ₦500
#   - 13 to 17        → ₦1000
#   - 18 and above    → ₦1500
#   - 60 and above    → ₦750 (senior discount)
# Write a program that takes an age and prints the ticket price.

age = 65   # test with different values: 3, 10, 16, 30, 65


```python
age = 65   # test with different values: 3, 10, 16, 30, 65

if age < 5:
    price = 0
elif age >= 60:
    price = 750
elif age >= 18:
    price = 1500
elif age >= 13:
    price = 1000
else:   # age 5 to 12
    price = 500

print("Ticket price: ₦", price)
```

# Q2. (10 marks)
# Given this list of temperatures (in Celsius) for 7 days,
# use a for loop to:
#   a) Print each temperature and label it "Hot" (>= 35), "Warm" (20–34), or "Cold" (< 20)
#   b) Count how many Hot, Warm, and Cold days there are
#   c) Print the counts at the end

temps = [38, 22, 17, 40, 29, 15, 36]

```python
temps = [38, 22, 17, 40, 29, 15, 36]

hot_count = 0
warm_count = 0
cold_count = 0

for temp in temps:
    if temp >= 35:
        print(temp, "→ Hot")
        hot_count += 1
    elif temp >= 20:
        print(temp, "→ Warm")
        warm_count += 1
    else:
        print(temp, "→ Cold")
        cold_count += 1

print("\nSummary:")
print("Hot days:", hot_count)
print("Warm days:", warm_count)
print("Cold days:", cold_count)
```

# Q4. (12 marks)
# Use nested loops and conditionals to print this pattern:
#
#   *
#   * *
#   * * *
#   * * * *
#   * * * * *
#
# Then below it, print only rows where the row number is odd (1, 3, 5).

```python
# First pattern
for i in range(1, 6):          # rows 1 to 5
    for j in range(i):         # print stars based on row number
        print("*", end=" ")
    print()                    # move to next line

print("\nOdd rows only:")

# Only odd rows (1, 3, 5)
for i in range(1, 6):
    if i % 2 == 1:             # check if row number is odd
        for j in range(i):
            print("*", end=" ")
        print()
```

# P7. Create a dictionary called contacts with at least 4 people (name → phone).
# Then:
#   a) Print all names using a loop
#   b) Add a new contact
#   c) Remove one contact using pop()
#   d) Check if "Ngozi" is in the contacts

```python
# create dictionary with 4 contacts
contacts = {
    "John": "08012345678",
    "Mary": "08087654321",
    "David": "08123456789",
    "Aisha": "07011223344"
}

# a) print all names using a loop
print("Contact names:")
for name in contacts:
    print(name)

# b) add a new contact
contacts["Peter"] = "09099887766"

# c) remove one contact using pop()
contacts.pop("Mary")

# d) check if "Ngozi" is in the contacts
if "Ngozi" in contacts:
    print("\nNgozi is in the contacts.")
else:
    print("\nNgozi is NOT in the contacts.")

# print updated dictionary
print("\nUpdated contacts:")
print(contacts)
```
